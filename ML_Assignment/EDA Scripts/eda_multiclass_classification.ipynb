### MNIST Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tensorflow.keras.datasets import mnist

# Load dataset
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Convert to DataFrame
num_samples = x_train.shape[0]
df = pd.DataFrame({'label': y_train})

# Display basic info
print(df.info())
print(df.describe())
print(df['label'].value_counts())

# Distribution of target variable (Digits 0-9)
sns.countplot(x='label', data=df, hue='label', palette='coolwarm', legend=False)
plt.show()

# Visualizing sample images
fig, axes = plt.subplots(1, 10, figsize=(15, 5))
for i in range(10):
    axes[i].imshow(x_train[i], cmap='gray')
    axes[i].axis('off')
plt.show()

# Save dataset (labels only since images are arrays)
df.to_csv("E:\Projects\Ml_ops\ML_Assignment\EDA Scripts\Datasets\mnist_labels.csv", index=False)
```
